```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df["duration_min"] = df["duration_ms"] / 60000
```

```python
df=df.drop_duplicates()
```

```python
print("Dataset After Transformation:")
print(df.head())
```

```python
# HANDLING NULL VALUES

#checking null values before handling
print("Null Values Before Handling:")
print(df.isnull().sum())
```

```python
## finding numerical columns
numeric_columns = df.select_dtypes(include="number").columns
```

```python
for column in numeric_columns:
        df[column] = df[column].fillna(df[column].mean())
```

```python
categorical_columns = df.select_dtypes(include=["object","str"]).columns
```

```python
for column in categorical_columns:
    if df[column].isnull().any():
        if not df[column].mode().empty:
            df[column] = df[column].fillna(df[column].mode()[0])

            ## checking null values
print("Null Values After Handling:")
print(df.isnull().sum())
```

```python
# INSIGHT GENERATION


top_songs = df.sort_values(
    by="popularity", ascending=False
).head(10)

print("Top 10 Popular Songs:")
print(top_songs[["track_name", "artists", "popularity"]])


artist_popularity = df.groupby("artists")[
    "popularity"
].mean().sort_values(ascending=False).head(10)

print("\nTop 10 Artists:")
print(artist_popularity)


genre_popularity = df.groupby("track_genre")[
    "popularity"
].mean().sort_values(ascending=False).head(10)

print("\nTop 10 Popular Genres:")
print(genre_popularity)


print("\nAverage Danceability:")
print(df["danceability"].mean())


print("\nAverage Energy:")
print(df["energy"].mean())


print("\nAverage Tempo:")
print(df["tempo"].mean())
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10,5))
plt.bar(top_songs["track_name"], top_songs["popularity"])
plt.title("Top 10 Popular Spotify Songs")
plt.xlabel("Song Name")
plt.ylabel("Popularity")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


plt.figure(figsize=(10,5))
genre_popularity.plot(kind="bar")
plt.title("Top 10 Popular Music Genres")
plt.xlabel("Genre")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


plt.figure(figsize=(8,5))
plt.scatter(df["danceability"], df["energy"])
plt.title("Danceability vs Energy")
plt.xlabel("Danceability")
plt.ylabel("Energy")
plt.tight_layout()
plt.show()
```
